# Evidence RAG deterministic benchmark

Set `DATASET_DIR` to the mounted Kaggle dataset exported by `evidence-rag kaggle-export`. The dataset includes the package source and content hashes. Synthetic fixture metrics are not human benchmark results.

```python
import sys
from pathlib import Path

DATASET_DIR = Path("/kaggle/input/evidence-rag-benchmark")
if not DATASET_DIR.is_dir():
    raise FileNotFoundError(f"Set DATASET_DIR to the exported dataset: {DATASET_DIR}")
sys.path.insert(0, str(DATASET_DIR))
```

```python
from evidence_rag.benchmark import run_benchmark, write_benchmark_report

report = run_benchmark(DATASET_DIR / "benchmark-manifest.json")
json_path, markdown_path = write_benchmark_report(
    report, Path("/kaggle/working/benchmark-results")
)
print(json_path)
print(markdown_path)
```
